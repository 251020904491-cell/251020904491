```python
n = int(input("Nhập số nguyên dương n: "))
tong = 0
for i in range(2, n + 1):
    if i % 2 == 0:
        tong += i
print("Tổng các số chẵn nhỏ hơn hoặc bằng n là:", tong)
```

```python
s = input("Nhập chuỗi: ")
words = s.split()
print("Số từ trong chuỗi:", len(words))
tu_dai_nhat = words[0]
for tu in words:
    if len(tu) > len(tu_dai_nhat):
        tu_dai_nhat = tu
print("Từ dài nhất là:", tu_dai_nhat)
```

```python
ds = list(map(int, input("Nhập danh sách các số nguyên: ").split()))
ds = list(set(ds))
ds.sort()
if len(ds) >= 2:
    so_lon_thu_2 = ds[-2]
    print("Danh sách sau xử lý:", ds)
    print("Số lớn thứ 2:", so_lon_thu_2)
else:
    print("Không đủ phân tử để tìm số lớn thứ 2")
```

```python
diem = list(map(float, input("danh sách điểm: ").split()))
diem_hop_le = [d for d in diem if 0 <= d <= 10]
diem_dat = [d for d in diem_hop_le if d >= 5]
diem_tb = sum(diem_dat)/len(diem_dat) if diem_dat else 0
so_dat = len(diem_dat)
print(f"trung bình: {diem_tb:.6f}")
print(f"đạt: {so_dat}")
```
